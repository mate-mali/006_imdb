```python
import pandas as pd
df = pd.read_csv("imdb_top250.csv")
print(df.head())
print(df.describe())
```

```python
# lista kolumn 
print(df.columns)
```

```python
# 1 - Tytuły filmów wyreżyserowanych przez Christophera Nolana.
udir = df.copy(deep=True)
udir['distinct director'] = df['director(s)'].str.split(',') #podizel dyrektorow po przecinkach
udir = udir.explode('distinct director') #podzielonych dyrektorow dodaj do osobnych rzedow dla tego samego filmu
udir['distinct director'] = udir['distinct director'].str.strip() #usun whitespaces przed i po nazie dyrektora

print("### 1 - Tytuły filmów wyreżyserowanych przez Christophera Nolana.")
for val in udir[udir['director(s)'] == 'Christopher Nolan']['title'].values: print(val)
```

```python
# 2 -Tytuły filmów, w których zagrał Tom Hardy.
df2 = df.copy(deep=True)
df2 = df2.set_index('actors') #zmiana indexu zeby uzyc filter na indexie

print("### 2 -Tytuły filmów, w których zagrał Tom Hardy.")
for val in df2.filter(like='Tom Hardy', axis = 0)['title'].values: print(val)
```

```python
# 10 reżyserów z największą liczbą filmów na liście. - alternatywa ktora uwzglednia duplikaty
top10_dir = udir.groupby('distinct director')['title'].count() #policz rzedy per unikatory dyrektor
top10_dir = top10_dir.sort_values(ascending=False)[:10] #wez top 10 wedlug posortowanej liczby filmow
print("### 3 - 10 reżyserów z największą liczbą filmów na liście.")
print(top10_dir) #udir i top_10 dir beda uzywane w nastepnym cwiczeniu
```

```python
# Tytuły filmów nakręconych przez 10 reżyserów z największą liczbą filmów na liście.
top10_dir_movies = pd.merge(left = udir, right = top10_dir, how='inner', left_on='distinct director', right_index = True)

print("### 4 - Tytuły filmów nakręconych przez 10 reżyserów z największą liczbą filmów na liście.")
print(top10_dir_movies['title_x'].values)
```

```python
# 9 aktorów występujących w największej liczbie filmów z listy.
actrs = df.copy(deep=True)
actrs['distinct actors'] = actrs['actors'].str.split(',')
actrs = actrs.explode('distinct actors')
actrs['distinct actors'] = actrs['distinct actors'].str.strip()
top9_act = actrs.groupby('distinct actors')['title'].count()
top9_act = top9_act.sort_values(ascending=False)[:9]

print("### 5 - 9 aktorów występujących w największej liczbie filmów z listy.")
print(top9_act)
```

```python
# Tytuły filmów z udziałem aktorów z największą liczbą filmów na liście.
top9_act_movies = pd.merge(left = actrs, right = top9_act, how='inner', left_on='distinct actors', right_index = True)

print("### 6 - Tytuły filmów z udziałem aktorów z największą liczbą filmów na liście.")
print(top9_act_movies['title_x'].values)
```

```python
# 5 najczęstszych "duetów" aktorskich (aktorów, którzy występowali razem w filmach z IMDB Top 250).
lact = actrs[['distinct actors', 'title']].copy(deep=True)
ract = actrs[['distinct actors', 'title']].copy(deep=True)

cjoin = pd.merge(lact, ract, left_on = 'title', right_on = 'title')             #crossjoin zeby stworzyc pary aktorow kazdy z kazdym per tytul
cjoin = cjoin[cjoin['distinct actors_x'] != cjoin['distinct actors_y']]         #   usuwamy pary gdzie aktor jest po obu stronach
cjoin['duo'] = cjoin['distinct actors_x'] + ", " + cjoin['distinct actors_y']   # tworzymy pierwszego stringa zeby potem z niego zrobic liste bo zapomnialem jak to zrobic w elegancji sposob
cjoin['duo'] = cjoin['duo'].str.split(', ')                                     #   dzielimy go z powrotem ale teraz juz mamy liste dwojek
cjoin['duo'] = cjoin['duo'].apply(lambda x: sorted(x))                          #    sortuj liste dwojek zeby usunac duplikaty
cjoin['duet of actors'] = cjoin['duo'].str.join(' + ')                          # kolejny klucz stringowy aleteraz ladny
duets = cjoin[['duet of actors', 'title']].drop_duplicates()                    # wez tylko te dwie kolumny i usun od razu duplikaty
popular_duos = duets.groupby('duet of actors')['title'].count()
popular_duos = popular_duos.sort_values(ascending=False)[:5] 

print('### 7 - 5 najczęstszych "duetów" aktorskich (aktorów, którzy występowali razem w filmach z IMDB Top 250).')
print(popular_duos)
```
