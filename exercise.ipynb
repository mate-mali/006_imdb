```python
# 1 - Tytuły filmów wyreżyserowanych przez Christophera Nolana.
print(df[df['director(s)'] == 'Christopher Nolan']['title'].values)
```

```python
# 2 -Tytuły filmów, w których zagrał Tom Hardy.
df2 = df.copy(deep=True)
df2 = df2.set_index('actors') #zmiana indexu zeby uzyc filter na indexie
print(df2.filter(like='Tom Hardy', axis = 0)['title'].values)
```

```python
# 10 reżyserów z największą liczbą filmów na liście. - alternatywa ktora uwzglednia duplikaty

dff = df 
dff['distinct director'] = dff['director(s)'].str.split(',')
dff= dff.explode('distinct director')
dff['distinct director'] = dff['distinct director'].str.strip()
#print(dff['distinct director'].to_csv('trim_check.csv'))
dff = dff.groupby('distinct director')['title'].count()
df_top10 = dff.sort_values(ascending=False)[:10]
print(df_top10)

```
